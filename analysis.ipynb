# Prime & Elite — исследовательский анализ данных
Учебные синтетические данные. Оценки соответствия не являются реальными результатами обслуживания.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
root=Path("data")
managers=pd.read_csv(root/"managers.csv")
clients=pd.read_csv(root/"clients.csv")
pairs=pd.read_csv(root/"training_pairs.csv")
assignments=pd.read_csv(root/"assignments.csv")
print("Менеджеры",managers.shape,"Клиенты",clients.shape,"Пары",pairs.shape,"Назначения",assignments.shape)
```

## Проверка качества и структуры

```python
for name,df in [("managers",managers),("clients",clients),("pairs",pairs),("assignments",assignments)]:
    print(name,"дубликатов:",df.duplicated().sum(),"пропусков:",df.isna().sum().sum())
    display(df.head(3))
```

```python
assert managers.manager_id.is_unique
assert clients.client_id.is_unique
assert (managers.initial_clients<=managers.capacity).all()
assert pairs.match_score.between(0,100).all()
assert set(assignments.client_id).issubset(set(clients.client_id))
print("Базовые проверки пройдены")
```

## Распределения

```python
fig,ax=plt.subplots(figsize=(7,4))
clients.segment.value_counts().plot.bar(ax=ax)
ax.set(title="Клиенты по сегментам",xlabel="Сегмент",ylabel="Количество")
plt.tight_layout();plt.show()
```

```python
fig,ax=plt.subplots(figsize=(7,4))
(managers.initial_clients/managers.capacity).hist(ax=ax,bins=15)
ax.set(title="Доля заполнения портфеля",xlabel="Нагрузка / ёмкость",ylabel="Менеджеров")
plt.tight_layout();plt.show()
```

```python
fig,ax=plt.subplots(figsize=(7,4))
pairs.match_score.hist(ax=ax,bins=30)
ax.set(title="Синтетическая оценка соответствия",xlabel="match_score",ylabel="Пар")
plt.tight_layout();plt.show()
```
